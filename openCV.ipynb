```python
#loading image
import cv2
image=cv2.imread('img.jpeg')

if image is None:
  print('error: image not found')
else:
  print("image loaded successfully")
```

```python
#1->coloured (by default)
#0-> grayscale
#-1-> unchanged
```

```python
from google.colab.patches import cv2_imshow
```

```python
#for displaying
#imshow, waitkey,destroyallwindows
cv2_imshow(image) #will show the image
cv2.waitKey(0) #screen pause ho jaye --> wait for a key
cv2.destroyAllWindows() #then band bhi ho jaye -> close the window
```

```python
cv2.imwrite("output.jpeg",image)
```
