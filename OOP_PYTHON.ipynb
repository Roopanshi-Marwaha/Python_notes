```python
class Atm:

  #constructor is a special function --> andar ke code ko execute karne ke liye u don't need to call it automatically gets called
  def __init__(self):
    self.pin=''
    self.balance=0
    print("constructor executed")
    self.menu()


  def menu(self):
    user_input=input("""
    Hi how can I help you?
    1. Press 1 to create pin
    2. Press 2 to change pin
    3. Press 3 to check balance
    4. Press 4 to withdraw
    5. Anything else to exit
    """)

    if user_input =='1':
      self.create_pin()
    elif user_input =='2':
      self.change_pin()
    elif user_input =='3':
      self.check_balance()
    elif user_input =='4':
      self.withdraw()
    else:
      exit()

  def create_pin(self):
    user_pin=input("enter your pin")
    self.pin=user_pin

    user_balance=int(input("enter balance"))
    self.balance=user_balance

    print('pin created successfully')
    self.menu()

  def change_pin(self):
    old_pin=input("enter old pin")
    if old_pin==self.pin:
      new_pin=input("enter new pin")
      self.pin=new_pin
      print("pin changed successfully")
    else:
      print("you entered wrong pin")
      self.menu()

  def check_balance(self):
    user_pin=input("enter your pin")
    if user_pin==self.pin:
      print("your balance is",self.balance)
    else:
      print("you entered wrong pin")
      self.menu()

  def withdraw(self):
    user_pin=input("enter your pin")
    if user_pin==self.pin:
      amount=int(input("enter the amount"))
      if amount<=self.balance:
        self.balance=self.balance-amount
        print("withdrawl successful")
      else:
        print("insufficient balance")
    else:
      print("you entered wrong pin")
      self.menu()

#variables ko hamesha constructor ke andar banaoge
#uske aage self lagaoge
#function jab bhi banaoge class ke andar ek paramter hona chahiye self karke
```

```python
obj=Atm()
print(type(obj)) #shows ki atm class ka object hai yeh
```

###WHY SELF?

```python
#golden rule of oop
#class ke andar ek data ya objects ko only class ka object access kar skata hai.
#class ke andar ka ek method bhi dusre method ko call nahi kar sakta
#so for this we use self so that one method can talk to another one using self
#as method --> class ke object (self) se baat karega --> another methods se baat karega, now both the methods can talk to each other
#self is called this in other languages
#self is just a convention you can use another name for it as well
```

```python
class Temp:
  def __init__(self):
    print('hello')
    print("self id",id(self))

obj=Temp()
print("obj  id",id(obj))
#we can see both points to same memory so self is nothing but our obj itself
```
