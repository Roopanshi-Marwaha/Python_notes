```python
class Atm:

  #constructor is a special function --> andar ke code ko execute karne ke liye u don't need to call it automatically gets called
  def __init__(self):
    self.pin=''
    self.balance=0
    print("constructor executed")
    self.menu()


  def menu(self):
    user_input=input("""
    Hi how can I help you?
    1. Press 1 to create pin
    2. Press 2 to change pin
    3. Press 3 to check balance
    4. Press 4 to withdraw
    5. Anything else to exit
    """)

    if user_input =='1':
      self.create_pin()
    elif user_input =='2':
      self.change_pin()
    elif user_input =='3':
      self.check_balance()
    elif user_input =='4':
      self.withdraw()
    else:
      exit()

  def create_pin(self):
    user_pin=input("enter your pin")
    self.pin=user_pin

    user_balance=int(input("enter balance"))
    self.balance=user_balance

    print('pin created successfully')
    self.menu()

  def change_pin(self):
    old_pin=input("enter old pin")
    if old_pin==self.pin:
      new_pin=input("enter new pin")
      self.pin=new_pin
      print("pin changed successfully")
    else:
      print("you entered wrong pin")
      self.menu()

  def check_balance(self):
    user_pin=input("enter your pin")
    if user_pin==self.pin:
      print("your balance is",self.balance)
    else:
      print("you entered wrong pin")
      self.menu()

  def withdraw(self):
    user_pin=input("enter your pin")
    if user_pin==self.pin:
      amount=int(input("enter the amount"))
      if amount<=self.balance:
        self.balance=self.balance-amount
        print("withdrawl successful")
      else:
        print("insufficient balance")
    else:
      print("you entered wrong pin")
      self.menu()

#variables ko hamesha constructor ke andar banaoge
#uske aage self lagaoge
#function jab bhi banaoge class ke andar ek paramter hona chahiye self karke
```

###WHY SELF?

```python
#golden rule of oop
#class ke andar ek data ya objects ko only class ka object access kar skata hai.
#class ke andar ka ek method bhi dusre method ko call nahi kar sakta
#so for this we use self so that one method can talk to another one using self
#as method --> class ke object (self) se baat karega --> another methods se baat karega, now both the methods can talk to each other
#self is called this in other languages
#self is just a convention you can use another name for it as well
#in case of multiple objects, self points to the current object
```

```python
class Temp:
  def __init__(self):
    print('hello')
    print("self id",id(self))

obj=Temp()
print("obj  id",id(obj))
#we can see both points to same memory so self is nothing but our obj itself
```

###CREATING FRACTIONS (NEW DATA TYPE)

```python
class Fraction:
  def __init__(self,x,y): #parametrized constructor
    self.num=x
    self.den=y

  def __str__(self): #print se by default idhar ata hai
    return '{}/{}'.format(self.num,self.den)

  def __add__(self,other): #names are convention--> 1st wala self hoga (3/4) and dusra wala other hoga (1/2)
  #automatically execute ho jayega jab plus sign hoga
    new_num=self.num*other.den+other.num*self.den
    new_den=self.den*other.den
    return '{}/{}'.format(new_num,new_den)

  def __sub__(self,other): #automatically execute ho jayega jab minus sign hoga
    new_num=self.num*other.den-other.num*self.den
    new_den=self.den*other.den
    return '{}/{}'.format(new_num,new_den)

  def __mul__(self,other): #automatically execute ho jayega jab * hoga
    new_num=self.num*other.num
    new_den=self.den*other.den
    return '{}/{}'.format(new_num,new_den)

  def convert_to_decimal(self):
    return self.num/self.den
```

```python
fr1=Fraction(3,8)
print(fr1) # as by default 3/4 nahi ata hai we have to use str which is a magic function
```

```python
fr2=Fraction(1,8)
print(fr1+fr2)
```

```python
print(fr1-fr2)
```

```python
print(fr1*fr2)
```

```python
fr1.convert_to_decimal()
```

### Write OOP classes to handle the following scenarios:

- A user can create and view 2D coordinates
- A user can find out the distance between 2 coordinates
- A user can find find the distance of a coordinate from origin
- A user can check if a point lies on a given line
- A user can find the distance between a given 2D point and a given line

```python
class Point:

  def __init__(self,x,y):
    self.x_cord=x
    self.y_cord=y

  def __str__(self):
    return '<{},{}>'.format(self.x_cord,self.y_cord)

  def euclidean_distance(self,other):
    return ((self.x_cord-other.x_cord)**2+(self.y_cord-other.y_cord)**2)**0.5

  def distance_from_origin(self):
    return ((self.x_cord)**2+(self.y_cord)**2)**0.5

class Line:
  def __init__(self,A,B,C):
    self.A=A
    self.B=B
    self.C=C

  def __str__(self):
    return '{}*x+{}*y+{}=0'.format(self.A,self.B,self.C)

  def point_on_line(self, point): #hamne point object banakr bheja hai main se
      if self.A*point.x_cord + self.B*point.y_cord + self.C == 0:
          return "lies on the line"
      else:
          return "does not lie on the line"

  def shortest_distance(self, point):
      return abs(self.A*point.x_cord + self.B*point.y_cord + self.C)/(self.A**2+self.B**2)**0.5
```

##How objects access attributes

```python
class Person:

  def __init__(self,name_input,country_input):
    self.name = name_input
    self.country = country_input

  def greet(self):
    if self.country == 'india':
      print('Namaste',self.name)
    else:
      print('Hello',self.name)
```

### Reference Variables

- Reference variables hold the objects
- We can create objects without reference variable as well
- An object can have multiple reference variables
- Assigning a new reference variable to an existing object does not create a new object

```python
# object without a reference
class Person:

  def __init__(self):
    self.name='Roopanshi Marwaha'
    self.gender='female'

p=Person() #Person() call karne se ek obj bana memory mei and ham p mei uska reference store karre hai --> p is not the obj it just contains the reference of the object created (it is a reference variable)
q=p #(aliasing) both are variable names pointing to same object
```

```python
# Multiple ref
print(id(p))
print(id(q))
```

```python
# change attribute value with the help of 2nd object
print(p.name)
print(q.name)
q.name='Raghav'
print(q.name) #ek mei change kare toh dusre mei bhi changes honge hi --> that is very risky and we should take care of it
print(p.name)
```
