###CREATING FRACTIONS (NEW DATA TYPE)

```python
class Fraction:
  def __init__(self,x,y): #parametrized constructor
    self.num=x
    self.den=y

  def __str__(self): #print se by default idhar ata hai
    return '{}/{}'.format(self.num,self.den)

  def __add__(self,other): #names are convention--> 1st wala self hoga (3/4) and dusra wala other hoga (1/2)
  #automatically execute ho jayega jab plus sign hoga
    new_num=self.num*other.den+other.num*self.den
    new_den=self.den*other.den
    return '{}/{}'.format(new_num,new_den)

  def __sub__(self,other): #automatically execute ho jayega jab minus sign hoga
    new_num=self.num*other.den-other.num*self.den
    new_den=self.den*other.den
    return '{}/{}'.format(new_num,new_den)

  def __mul__(self,other): #automatically execute ho jayega jab * hoga
    new_num=self.num*other.num
    new_den=self.den*other.den
    return '{}/{}'.format(new_num,new_den)

  def convert_to_decimal(self):
    return self.num/self.den
```

```python
fr1=Fraction(3,8)
print(fr1) # as by default 3/4 nahi ata hai we have to use str which is a magic function
```

```python
fr2=Fraction(1,8)
print(fr1+fr2)
```

### Write OOP classes to handle the following scenarios:

- A user can create and view 2D coordinates
- A user can find out the distance between 2 coordinates
- A user can find find the distance of a coordinate from origin
- A user can check if a point lies on a given line
- A user can find the distance between a given 2D point and a given line

```python
class Point:

  def __init__(self,x,y):
    self.x_cord=x
    self.y_cord=y

  def __str__(self):
    return '<{},{}>'.format(self.x_cord,self.y_cord)

  def euclidean_distance(self,other):
    return ((self.x_cord-other.x_cord)**2+(self.y_cord-other.y_cord)**2)**0.5

  def distance_from_origin(self):
    return ((self.x_cord)**2+(self.y_cord)**2)**0.5

class Line:
  def __init__(self,A,B,C):
    self.A=A
    self.B=B
    self.C=C

  def __str__(self):
    return '{}*x+{}*y+{}=0'.format(self.A,self.B,self.C)

  def point_on_line(self, point): #hamne point object banakr bheja hai main se
      if self.A*point.x_cord + self.B*point.y_cord + self.C == 0:
          return "lies on the line"
      else:
          return "does not lie on the line"

  def shortest_distance(self, point):
      return abs(self.A*point.x_cord + self.B*point.y_cord + self.C)/(self.A**2+self.B**2)**0.5
```

### Reference Variables

- Reference variables hold the objects
- We can create objects without reference variable as well
- An object can have multiple reference variables
- Assigning a new reference variable to an existing object does not create a new object

```python
# object without a reference
class Person:

  def __init__(self):
    self.name='Roopanshi Marwaha'
    self.gender='female'

p=Person() #Person() call karne se ek obj bana memory mei and ham p mei uska reference store karre hai --> p is not the obj it just contains the reference of the object created (it is a reference variable)
q=p #(aliasing) both are variable names pointing to same object
```

```python
class Atm:

  __counter = 1

  # constructor(special function)->superpower ->
  def __init__(self):
    print(id(self))
    self.pin = ''
    self.__balance = 0
    self.cid = Atm.__counter
    Atm.__counter = Atm.__counter + 1
    #self.menu()

  # utility functions --> chota mota kaam karne wale which don't need to make objects.
  @staticmethod #add this decorator so now we know that it is static
  def get_counter(): #bina self ke so now we access it like Atm.get_counter() and not simply c1.get_counter() and then c1 will be passed as one paramter
    return Atm.__counter


  def get_balance(self):
    return self.__balance

  def set_balance(self,new_value):
    if type(new_value) == int:
      self.__balance = new_value
    else:
      print('beta bahot maarenge')

  def __menu(self):
    user_input = input("""
    Hi how can I help you?
    1. Press 1 to create pin
    2. Press 2 to change pin
    3. Press 3 to check balance
    4. Press 4 to withdraw
    5. Anything else to exit
    """)

    if user_input == '1':
      self.create_pin()
    elif user_input == '2':
      self.change_pin()
    elif user_input == '3':
      self.check_balance()
    elif user_input == '4':
      self.withdraw()
    else:
      exit()

  def create_pin(self):
    user_pin = input('enter your pin')
    self.pin = user_pin

    user_balance = int(input('enter balance'))
    self.__balance = user_balance

    print('pin created successfully')

  def change_pin(self):
    old_pin = input('enter old pin')

    if old_pin == self.pin:
      # let him change the pin
      new_pin = input('enter new pin')
      self.pin = new_pin
      print('pin change successful')
    else:
      print('nai karne de sakta re baba')

  def check_balance(self):
    user_pin = input('enter your pin')
    if user_pin == self.pin:
      print('your balance is ',self.__balance)
    else:
      print('chal nikal yahan se')

  def withdraw(self):
    user_pin = input('enter the pin')
    if user_pin == self.pin:
      # allow to withdraw
      amount = int(input('enter the amount'))
      if amount <= self.__balance:
        self.__balance = self.__balance - amount
        print('withdrawl successful.balance is',self.__balance)
      else:
        print('abe garib')
    else:
      print('sale chor')
```

##### Points to remember about static

- Static attributes are created at class level.
- Static attributes are accessed using ClassName.
- Static attributes are object independent. We can access them without creating instance (object) of the class in which they are defined.
- The value stored in static attribute is shared between all instances(objects) of the class in which the static attribute is defined.

```python
class Lion:
  __water_source="well in the circus"

  def __init__(self,name, gender):
      self.__name=name
      self.__gender=gender

  def drinks_water(self):
      print(self.__name,
      "drinks water from the",Lion.__water_source)

  @staticmethod
  def get_water_source():
      return Lion.__water_source

simba=Lion("Simba","Male")
simba.drinks_water()
print( "Water source of lions:",Lion.get_water_source())
```
