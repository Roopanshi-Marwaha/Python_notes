```python
#guessing game
#0-100 random number select example 50
#now user has to guess

import random
jackpot=random.randint(1,100)
guess=int(input('guess karo'))

while guess!=jackpot:
  if guess<jackpot:
    print('guess higher')
  elif guess>jackpot:
    print('guess lower')
  else:
    print('correct guess')

  guess=int(input('guess karo'))

else:
  print('correct guess')
#har baar loop ke bahara nikal ke galati bata rahe hai (loop ke anadar) and bahar niklakr 2nd chance de rahe hai
#agar ham loop ke andar hai matlab hamne galat hi choose kara hai
#as sahi guess hota toh loop se bahar hote
```

##SEQUENCE SUM

```python
#program to solve 1/1! + 2/2! + 3/3!
import math
n=int(input('enter n'))
sum=0
for i in range(1,n+1):
  sum=sum+i/math.factorial(i)
print(sum)
```

```python
#pattern
# *
# **
# ***
n=int(input('enter n'))
for i in range(1,n+1):
  for j in range(1,i+1):
    print("*",end="")
  print() #to change the line
```

```python
#pattern
# 1
# 121
# 12321
# 1234321
n=int(input('enter n: '))
for i in range (1,n+1):
  for j in range(1,i+1):
    print(j,end='')
  for k in range(i-1,0,-1): #0 not included toh 1 pr ruk jayega --> additional loop to be executed
    print(k,end='')
  print()
```
