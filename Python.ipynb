#Literals

```python
#jo bhi raw value ko ham store kar rahe hai (variable ko jo value milta hai)
# Integer Literals

a = 0b101010  # Binary Literal (0b python ko bata raha hai ki yeh binary hai)
b = 100       # Decimal Literal (normal)
c = 0o310     # Octal Literal (0o tells octal)
d = 0x12c     # Hexadecimal Literal (0x tells hexadecimal)

# Float Literals

float_1 = 10.5
float_2 = 1.5e2 #1.5*10^2 -->bade
float_3 = 1.5e-3 #1.5*10^-3 -->chote numbers

# Complex Literal

x = 3.14j

print(a, b, c, d)
print(float_1, float_2, float_3)
print(x)
print(x.real,x.imag) #how to deal with complex, real->real, imaginary-> imag
```
