### Difference between Lists and Tuples

```python
# Syntax (lists-> square brackets, tuples-> ())
# Mutability (lists-> mutable tuples-> immutable)
# Speed (tuples are fasters than lists, generally, immutable are faster and takes lesser space than mutable data types)
# Memory (tuples take lesser memory in comparsion to tuples)
# Built in functionality (lists have more built in functions than tuples)
# Error prone (list is more error prone)
# Usability (college wala example)
```

```python
#speed difference
import time

L = list(range(100000000))
T = tuple(range(100000000))

start = time.time()
for i in L:
  i*5
print('List time',time.time()-start)

start = time.time()
for i in T:
  i*5
print('Tuple time',time.time()-start)
```

```python
#memory difference
import sys

L = list(range(1000))
T = tuple(range(1000))

print('List size',sys.getsizeof(L))
print('Tuple size',sys.getsizeof(T))
```

```python
#*args
#allows us to pass a variable number of non keyword arguments to a functions

def multiply(*args): #asterisk --> variable (it internally creats a tuple and keeps adding them to it)
  product=1
  for i in args:
    product=product*i
  print(args)
  return product
```

```python
multiply(2,3,4,6)
```

```python
#**kwargs allows us to pass any number of keyword arguments.
#keyword arguments mean that they contain a key-value pair, like a python dictionary.
# **kwargs directly normal positional arguments ke liye use nahi hota.

def display(**kwargs): #** means -> key, value pairs ane wale hai
  for (key,value) in kwargs.items(): #items-> har key,value pairs ko alag kar deta hai
    print(key,'->',value)
```

```python
display(india='delhi',srilanka='colombo',nepal='kathmandu',pakistan='islamabad')
```

### Lambda Function

A lambda function is a small anonymous function.

A lambda function can take any number of arguments, but can only have one expression.

image.png

```python
# x -> x^2
lambda x:x**2
```

```python
# x,y -> x+y
a = lambda x,y:x+y
a(5,2)
```

#### Diff between lambda vs Normal Function

- No name
- lambda has no return value(infact,returns a function)
- lambda is written in 1 line
- not reusable

Then why use lambda functions?<br>
**They are used with HOF**


```python
# check if a string has 'a'
a = lambda s:'a' in s
a('hello')
```

```python
# odd or even
a = lambda x:'even' if x%2 == 0 else 'odd'
a(6)
```
